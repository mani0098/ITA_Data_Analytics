#### 1. Query the public [{JSON} Placeholder](https://jsonplaceholder.typicode.com/) API using the GET method to obtain:

- List of posts (/posts)

- User list (/users)

- Task list (/all)

```python
import requests
import pandas as pd
import time
```

```python
# Defining the URLs to be called
urlPosts = "https://jsonplaceholder.typicode.com/posts"
urlUsers = "https://jsonplaceholder.typicode.com/users"
urlTodos = "https://jsonplaceholder.typicode.com/todos"

# Using the GET method to call the APIs and store the responses in variables
respPosts = requests.get(urlPosts)
respUsers = requests.get(urlUsers)
respTodos = requests.get(urlTodos)
```

#### 2. Display on screen:

- The total amount of each resource.
- The status code of each request.

```python
# Extracting the data to JSON format
dataPosts = respPosts.json()
dataUsers = respUsers.json()
dataTodos = respTodos.json()

# Showing the results
print("Query (GET)")
print(f"Posts -> status code: {respPosts.status_code} | total amount: {len(dataPosts)}")
print(f"Users -> status code: {respUsers.status_code} | total amount: {len(dataUsers)}")
print(f"Todos -> status code: {respTodos.status_code} | total amount: {len(dataTodos)}")
```

#### 3. Make a request to a non-existent post to get a 404 error and display the status code received.

```python
false_url = "https://jsonplaceholder.typicode.com/posts/99999"

# Request to a non-existent post to check the response code
resp404 = requests.get(false_url)

print("Request to a non-existent resource")
print(f"Received status code: {resp404.status_code}") 
```

#### 4. Make a POST request to create a new dummy post. Include the title, message body, and a userId.

- Show the JSON response.
- Show status code.

```python
# Creating the data for the new publication
new_post = {
    "title": "Sprint 12",
    "body": "It's a body message created in Python.",
    "userId": 4
}

# Sending the POST request
response_post = requests.post("https://jsonplaceholder.typicode.com/posts", json=new_post)

print("JSON Response:")
print(response_post.json())
print("Status Code:", response_post.status_code)
```

#### 5. Make a PATCH request to partially modify an existing publication.

- Show the JSON response.
- Show status code.

```python
title_modification = {
    "title": "Updated title with PATCH"
}

response_patch = requests.patch("https://jsonplaceholder.typicode.com/posts/1", json=title_modification)

print("JSON Response:")
print(response_patch.json())
print("Status Code:", response_patch.status_code)
```

#### 6. Make a DELETE request on a post.

- Show the JSON response.
- Show status code.

```python
response_delete = requests.delete("https://jsonplaceholder.typicode.com/posts/1")

print("JSON Response:")
print(response_delete.json())
print("Status Code:", response_delete.status_code)
```

1. `/houses`: Returns the list of houses.

**Filters and optional parameters:** I noticed that the API allows filtering and searching data using parameters in the URL (query parameters). For example, in the `/houses` endpoint, I can use:

**JSON Format:**
Yes, the API returns its responses in JSON format. A list request returns a JSON array, while a request for one specific resource returns a JSON object. The response content type is `application/json`.

#### 3. Make a simple GET request:

- Displays the response status code.
- It clearly prints some of the fields in the JSON response.

```python
url_API = "https://anapioficeandfire.com/api/houses"

parameters = {
    "name": "House Stark of Winterfell"
}

# GET request
response = requests.get(url_API, params=parameters)

# Display the status code
print(f"Status Code: {response.status_code}")

# Convert the response to JSON
houses = response.json()

# Print selected fields
for house in houses:
    print(f"Name: {house['name']}")
    print(f"Region: {house['region']}")
    print(f"Words: {house['words']}")
```

#### 4. Convert the data to a pandas DataFrame:

- Displays the first rows of the DataFrame.

```python
# Convert the JSON data into a DataFrame
houses_df = pd.DataFrame(houses)

filtered_houses_df = houses_df[["name", "region", "words", "titles", "founded", "ancestralWeapons"]]

# Display the first rows
filtered_houses_df.head()
```

## ⭐⭐⭐ **Level 3** 

[Open Data Barcelona](https://opendata-ajuntament.barcelona.cat/ca/desenvolupadors#APIS) API

#### 1. Use the Open Data BCN API to:

- Search for a dataset of your interest using package_search and package_show.

```python
# Search for datasets with package_search
base_url = "https://opendata-ajuntament.barcelona.cat/data/api/action"

search_parameters = {
    "q": "bicing",
    "rows": 5
}

search_response = requests.get(
    f"{base_url}/package_search",
    params=search_parameters,
    timeout=20
)

print(f"Status Code: {search_response.status_code}")

search_response.raise_for_status()

search_json = search_response.json()
datasets = search_json["result"]["results"]

print(f"Datasets found: {search_json['result']['count']}\n")

for dataset in datasets:
    print(f"Title: {dataset['title']}")
    print(f"Dataset ID: {dataset['name']}")
    print()
```

```python
# Examine one dataset with package_show

# Select the first dataset found
dataset_id = datasets[0]["name"]

show_parameters = {
    "id": dataset_id
}

show_response = requests.get(
    f"{base_url}/package_show",
    params=show_parameters,
    timeout=20
)

print(f"Status Code: {show_response.status_code}")

show_response.raise_for_status()

dataset_details = show_response.json()["result"]

print(f"Title: {dataset_details['title']}")
print(f"Dataset ID: {dataset_details['name']}")
print(f"Description: {dataset_details.get('notes', 'No description')}")
print(f"Number of resources: {len(dataset_details['resources'])}")
```

#### 2. From the results obtained, select one that has resources available in CSV or JSON.

```python
# Select the first dataset with a CSV or JSON resource
selected_dataset = None
selected_resource = None

for dataset in datasets:
    for resource in dataset.get("resources", []):
        resource_format = resource.get("format", "").upper()

        if resource_format in ["CSV", "JSON"]:
            selected_dataset = dataset
            selected_resource = resource
            break

    if selected_dataset is not None:
        break

# Display the selected dataset and resource
if selected_dataset:
    print(f"Dataset: {selected_dataset['title']}")
    print(f"Dataset ID: {selected_dataset['name']}")
    print(f"Resource format: {selected_resource['format']}")
    print(f"Resource URL: {selected_resource['url']}")
else:
    print("No CSV or JSON resources were found.")
```

#### 3. With the resource_id of the selected resource, perform a query using datastore_search to:

- Retrieve at least 100 records.

```python
# Selecting only resources whose datastore_active value is True
selected_dataset = None
selected_resource = None

for dataset in datasets:

    # Get complete dataset information
    response = requests.get(
        f"{base_url}/package_show",
        params={"id": dataset["name"]},
        timeout=20
    )

    dataset_details = response.json()["result"]

    for resource in dataset_details["resources"]:

        resource_format = resource.get("format", "").upper()
        datastore_active = resource.get("datastore_active", False)

        if resource_format in ["CSV", "JSON"] and datastore_active:
            selected_dataset = dataset_details
            selected_resource = resource
            break

    if selected_resource is not None:
        break
```

```python
# Check if a suitable resource was found
if selected_resource:
    print(f"Dataset: {selected_dataset['title']}")
    print(f"Format: {selected_resource['format']}")
    print(f"Resource ID: {selected_resource['id']}")
    print(f"DataStore active: {selected_resource['datastore_active']}")
else:
    print("No DataStore-enabled CSV or JSON resource was found.")
```

```python
# Retrieving 100 records
datastore_response = requests.get(
    f"{base_url}/datastore_search",
    params={
        "resource_id": selected_resource["id"],
        "limit": 100
    },
    timeout=20
)

print(f"Status Code: {datastore_response.status_code}")

datastore_response.raise_for_status()

datastore_json = datastore_response.json()
records = datastore_json["result"]["records"]

print(f"Records retrieved: {len(records)}")
```

#### 4. Convert the results to a DataFrame.

```python
df = pd.DataFrame(records)

df.head()
```

#### 5. Save the DataFrame to a .csv file.

```python
# Save the DataFrame as a CSV file
df.to_csv("Bicing_August18_March19.csv", index=False)

print("The file has been saved successfully.")
```
