#### 3.
- Merge the Fills and No Fills columns into a single column, using the .apply() method and defining a function that solves the problem. The new column should be called "Fills" and take the values ​​"Sí" or "No".

````python
def column_merger(row):
    '''
    A function to later merge two columns with a new definition
    '''
    if pd.notnull(row['Fills']) and pd.isnull(row['No Fills']):
        return 'Sí'
    elif pd.notnull(row['No Fills']) and pd.isnull(row['Fills']):
        return 'No'
    else:
        return 'Anomaly Detected!'


employee_survey['Fills_merged'] = employee_survey.apply(column_merger, axis=1)
````

````python
# Deleting the existing column and renaming the new one
employee_survey.drop(columns=["Fills", "No Fills"], inplace=True)
employee_survey.rename(columns={"Fills_merged": "Fills"}, inplace=True)
````

#### 1.

Create a function that takes a dataframe as an input parameter.

The function should automatically create (and export) a chart for each column in the dataframe. For example:

- a histogram/boxplot if the variable is numeric
- some bars of the most frequent values ​​if it is categorical
- some bars of the most frequent years if the data is in date format.

The idea is to create a function that works for **any** dataframe, not just the one we've worked with so far.

Show the result of the function on one of the example datasets contained in the seaborn package. For example, *iris*, *penguins*, or *titanic*.

**Keep in mind that in the next sprint you will work exclusively with graphics. The goal of this exercise is not to create very elaborate graphics, but to solve a need quickly and automatically.**

````python
def chart_plotter(df):
    for col in df.columns:
        series = df[col].dropna()

        plt.figure(figsize=(9,5))

        if is_bool_dtype(series):                            # Boleans
            series.value_counts().plot(kind='bar')
            plt.title(f'The most frequent values of {col}')
            plt.xticks(rotation=45)

        elif is_numeric_dtype(series):                       # Numerics
            sns.histplot(series, kde=True, color='purple')
            plt.title(f'Histogram of {col}')

        elif is_datetime64_any_dtype(series):                # Date & Time
            series.dt.year.value_counts().sort_index().plot(kind='bar', color='orange', edgecolor='red')
            plt.title(f'The most frequent years in {col}')

        else:                                                # None of the above
            series.value_counts().head(10).plot(kind='bar', color='#DAF7A6', edgecolor='#BB8FCE')
            plt.title(f'The most frequent values of {col}')
            plt.xticks(rotation=45)

        plt.tight_layout()
        file_name = f"{col}_chart.png"
        plt.savefig(file_name)
        plt.close()
        print(f"{file_name} exported successfully!")
````

````python
iris = sns.load_dataset("iris")
chart_plotter(iris)
````

````python
penguins = sns.load_dataset("penguins")
chart_plotter(penguins)
````

````python
titanic = sns.load_dataset("titanic")
chart_plotter(titanic)
````

#### 2.
Load the file matriu_distancies.xlsx into pandas, so that the row names and column names are those of the cities. Delete "Las Palmas de Gran Canaria" and "Palma" so that we can make the trip by car.

Source: [Best Routes](https://espana.mejoresrutas.com/tabla-de-distancias/es/)

We are interested in visiting all the main cities in Spain while traveling the shortest possible distance.

**You don't have to do it optimally, we are interested in you developing a reasonable solution using the tools you currently have.**

For example, a simple (but not optimal) approach would be to always go to the nearest city that we haven't visited yet.

Write a function that, given a distance matrix and the origin city, returns a route of minimum length, with the order of visits. Also, gives the total distance traveled. 

(EXTRA) From which city would the route be shortest with the proposed algorithm?

````python
# Reading the Excel file
df_distance = pd.read_excel("matriu_distancies.xlsx", index_col=0)

# Removing the desired cities
outlier_cities = ["Las Palmas de Gran Canaria", "Palma"]
df_distance_pruned = df_distance.drop(columns = outlier_cities, index = outlier_cities, errors = "ignore")

# Changing self-distance to 0
df_distance_pruned = df_distance_pruned.fillna(0)
df_distance_pruned
````

````python
def shortest_path(dist_matrix, home_city):
    """
    Finding a route using the Nearest Neighbor algorithm.
    """

    not_visited = set(dist_matrix.index)

    path = [home_city]
    not_visited.remove(home_city)

    total_distance = 0
    current_city = home_city

    while not_visited:

        # Nearest unvisited city
        next_city = (dist_matrix.loc[current_city, list(not_visited)].idxmin())

        total_distance += dist_matrix.loc[current_city, next_city]

        path.append(next_city)
        not_visited.remove(next_city)
        current_city = next_city

    return path, total_distance
````

````python
# Testing the function by a random city
route, distance = shortest_path(df_distance_pruned, "Murcia")

print("Route:")
print(" > ".join(route))

print("\nTotal distance:", distance, "km")
````

````python
# Finding the origin city for the shortest route
distance_order = []

for city in df_distance_pruned.index:

    _, distance = shortest_path(df_distance_pruned, city)

    distance_order.append({"Origin City": city, "Covered Distance (km)": int(distance)})

final_distance_order = (pd.DataFrame(distance_order).sort_values("Covered Distance (km)").reset_index(drop=True))

print(final_distance_order)
print("\n")
print(f"The shortest route is to start from {final_distance_order.loc[0, 'Origin City']}:")
trajectory, displacement = shortest_path(df_distance_pruned, final_distance_order.loc[0, 'Origin City'])
print(" > ".join(trajectory))
print(f"Total covered distance: {displacement:.0f} km")
````
