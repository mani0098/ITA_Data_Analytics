#### 1. Body Mass Index calculator
- Write a function that calculates the BMI entered by the user, that is, whoever runs it will have to enter this data. You can get more information about its calculation at: [Body Mass Index What is BMI and how is it calculated](https://muysalud.com/salud/indice-de-masa-corporal-imc-que-es-y-como-se-calcula/).
- The function must classify the result into its respective categories.

*Tip: Try validating the data beforehand, so that it sends a warning message if the data entered by the user is not in the appropriate format or does not take reasonable values.*

````python
def bmi_calculation(weight, height):
    """
    Calculation of BMI using weight (kilograms) and height (meters) using the formula: weight/height²
    """
    return round(weight / pow(height, 2), 2)

def bmi_category(bmi):
    """
    Assigning calculated BMI to its corresponding category
    """
    if bmi > 30:
        category = "Obese"
    elif bmi >= 25: 
        category = "Overweight"
    elif bmi >= 18.5:
        category = "Normal Weight"
    else:
        category = "Underweight"
    
    return category
````

#### 3. Word counter for a text
Write a function that, given a text, displays the number of times each word appears. Try to handle all possible cases that cause the program not to work correctly.

(EXTRA): What is the average word length of the text you wrote? “Hello, how are you?” should return (4+3+2) / 3 = 3

````python
import re
````

````python
def word_count(text):
    """
    Counting words in a sentence, excluding numbers and characters
    """    
    text = text.lower()
    text = re.sub(r"[^\w\s]", " ", text)
    #text = re.findall(r"[a-zA-Z]+", text)
    #words = text.split()
    words = [word for word in text.split() if word.isalpha()]
    count = {}

    for word in words:
        #word = word.strip('.,;:¡!¿?()@$%&~"')
        count[word] = count.get(word, 0) + 1
        
    # Average length
    if not words:
        average = 0.0         
    else:
        length = [len(p) for p in words]
        average = round(sum(length) / len(length), 2)

    return count, average
````

#### 4. Reverse dictionary (with possibility of duplicates)
It turns out that the client has a very old survey stored in a dictionary and needs the results in reverse, i.e., by swapping the keys and values. The values ​​and keys in the original dictionary are unique; if this is not the case, the function should print a warning message, along with a list of the values ​​associated with the repeated key.

Look at the following examples to see the expected output.

![alt text](image.png)

````python
def reverse_dictionary(dictionary):
    """
    Creating a new dictionary that converts key to value and value(s) to key from the inserted dictionary
    """
    if not isinstance(dictionary, dict):
        print("Error: Input must be in dictionary format.")

    reversed_dictionary = {}

    for key, value in dictionary.items():
        # If the value already exists in the new dictionary, it is added as a duplicate.
        if value in reversed_dictionary:
            print(f"Warning! The value '{value}' already exists.")
            
            # If there is no list, it is created.
            if not isinstance(reversed_dictionary[value], list):
                reversed_dictionary[value] = [reversed_dictionary[value], key]
            else:
                # In case there is a list, we only add the new key.
                reversed_dictionary[value].append(key)
        else:
            # If the value doesn't exist, it is added.
            reversed_dictionary[value] = key
    
    return reversed_dictionary
````

#### 1. Counter and sorter of words in a text
The client was happy with the word counter, but now he wants to read TXT files and calculate the frequency of each word ordered within the usual dictionary entries according to the letter they begin with, that is, the keys must go from A to Z and within A we must go from A to Z. For example, for the file "tu_me_quieres_blanca.txt" the expected output would be:

![alt text](image2.png)

````python
from collections import defaultdict
import string
import pprint
````

````python
def dictionary_style_word_count(text):
    """
    Creating a dictionary that counts all the words of a text and sorts them alphabetically and frequency-wise
    """

    words = text.lower().split()
    count = defaultdict(int)

    for word in words:
        word_cleaning = word.strip(string.punctuation)
        if word_cleaning:
            count[word_cleaning] += 1

    final_result = {}

    # Extracting the first letter to sort
    for word, freq in count.items():
        letter = word[0]

        if letter not in final_result:
            final_result[letter] = {}

        # Adding the word to the dictionary
        final_result[letter][word] = freq  

    # Sorting the words
    for letter in final_result:
        final_result[letter] = dict(sorted(final_result[letter].items()))

    # Sorting based on the letters
    return dict(sorted(final_result.items()))  
````

````python
with open(r"tu me quieres blanca.txt", encoding="utf-8") as file:
    song = file.read()

pprint.pprint(dictionary_style_word_count(song))
````

#### 2. Data type conversion
The client receives a list of data and needs to generate two lists, the first where all the elements that could be converted to floats will be and the other where the elements that could not be converted are. Example of the list that the client receives:

![alt text](image3.png)

````python
def numerically_separator(input_data):
    """
    Separating list elements under the condition of being a number
    """

    convertible = []
    not_convertible = []

    for item in input_data:

        # If it's a list, a tuple, or a set
        if isinstance(item, (list, tuple, set)):
            for subitem in item:
                try:
                    convertible.append(float(subitem))
                except:
                    not_convertible.append(subitem)

        # If it's a single element
        else:
            try:
                convertible.append(float(item))
            except:
                not_convertible.append(item)

    return convertible, not_convertible
````

#### 1. Password generator
Explore the operation of the random module of the numpy library. [Random module in NumPy](https://www.cybrosys.com/blog/random-module-in-numpy)

At this point, the client has detected a problem with the passwords used by their employees. Asdf1234, birthdays or similar. To solve this, they have commissioned us to create a Python function that generates more secure passwords. The function must depend on the following parameters:

- length (int): Password length
- capital letters (bool = True): Whether capital letters should appear
- lowercase (bool = True): Whether lowercase letters should appear
- numbers (bool = True): Whether numbers should appear
- signs (bool = False): If special characters should appear (,-$? or similar)

So, if we execute the function as follows:

```python
create_password(10, True, True, True, True)
```
We should obtain an output (which we must be able to save) like this:

*9Er,5Vn8P$*

Make sure that all criteria are met, and that these passwords are truly random.

(EXTRA) Explore how we could make the function automatically copy the password to the computer's clipboard (as if we had selected it and done ctrl+copy).

````python
import numpy as np
import string
import pyperclip   # To copy to the clipboard

def create_password(length, capital_letters=True, lowercase=True, numbers=True, signs=False):
    """
    Generating a random password with an arbitrary length
    """
    
    chars = ""
    password = []    
    required = sum([capital_letters, lowercase, numbers, signs])
    
    if required == 0:
        raise ValueError("Error: Choose at least one character type.")
    if length < required:
        raise ValueError(f"Error: Password must have at least {required} characters.")

    if capital_letters:
        chars += string.ascii_uppercase
        password.append(np.random.choice(list(string.ascii_uppercase)))

    if lowercase:
        chars += string.ascii_lowercase
        password.append(np.random.choice(list(string.ascii_lowercase)))

    if numbers:
        chars += string.digits
        password.append(np.random.choice(list(string.digits)))

    if signs:
        chars += string.punctuation
        password.append(np.random.choice(list(string.punctuation)))

    # Completing the remaining desired length
    remaining = length - len(password)
    password += list(np.random.choice(list(chars), remaining))

    # Combining all types of characters
    np.random.shuffle(password)

    final_pass = "".join(password)

    pyperclip.copy(final_pass)

    return final_pass
````

#### 2. Simple data processing
A coworker has asked us for a favor, taking advantage of the fact that she knows we are learning to program. She has a history of Catalan football matches in a file, where she has stored the names of the teams and the results. She needs us to process the data automatically, to extract the results she needs.

Use the file "historic_partits.txt"

You need a program that returns:

- The total number of goals scored by each team.
- The name of the top-scoring team.
- The name of the team with the most goals scored
- The overall standings (each win: 3 pts, draw 1 pts, defeat 0 pts)

For reference, the teams with the most points would be the following:

![alt text](image4.png)

You can use the format that is most convenient for you.

````python
def result_analysis(file_name):
    """
    Analyzing the results in a TXT format to extract the final standing table
    """
    teams = {}

    with open(file_name, "r", encoding="utf-8") as file:
        for line in file:
            try:
                team_A, result, team_B = line.strip().split('\t')   # team_A: home team & team_B: away team
            except ValueError:
                continue
            
            team_A_GF, team_B_GF = map(int, result.split('-'))
            
            # Defining and shaping the final standing table
            for team in [team_A, team_B]:
                teams.setdefault(
                        team,
                        {
                        'MP': 0,
                        'W': 0, 
                        'D': 0, 
                        'L': 0, 
                        'GF': 0, 
                        'GA': 0,
                        'GD': 0, 
                        'Pts': 0
                        }
                    )

            # Counting scored and conceded goals of each team
            teams[team_A]['GF'] += team_A_GF
            teams[team_A]['GA'] += team_B_GF
            teams[team_B]['GF'] += team_B_GF
            teams[team_B]['GA'] += team_A_GF

            # Counting the number of matches played
            teams[team_A]['MP'] += 1
            teams[team_B]['MP'] += 1

            # Counting the points in each match played
            if team_A_GF > team_B_GF:
                teams[team_A]['Pts'] += 3
                teams[team_A]['W'] += 1
                teams[team_B]['L'] += 1
            elif team_A_GF < team_B_GF:
                teams[team_B]['Pts'] += 3
                teams[team_B]['W'] += 1
                teams[team_A]['L'] += 1
            else:
                teams[team_A]['Pts'] += 1
                teams[team_B]['Pts'] += 1
                teams[team_A]['D'] += 1
                teams[team_B]['D'] += 1
                
    for team in teams:
        teams[team]["GD"] = teams[team]["GF"] - teams[team]["GA"]

    return teams
````

````python
import pandas as pd

try:
    league_stat = result_analysis("historic partits.txt")
except FileNotFoundError as e:
    print(e)

# Creating a DataFrame to show the final result
final_standing = pd.DataFrame.from_dict(league_stat, orient='index')

league_table = [
    'MP', 
    'W', 
    'D', 
    'L', 
    'GF', 
    'GA',
    'GD',
    'Pts'
]
final_standing = final_standing[league_table]

# Sorting the table based on points, goal difference, and goals scored (descending)
final_standing = final_standing.sort_values(by = ["Pts", "GD", "GF"], ascending = [False, False, False])

most_offensive_team = final_standing['GF'].idxmax()
most_goals_scored = final_standing['GF'].max()
most_defensive_team = final_standing['GA'].idxmax()
most_goals_conceded = final_standing['GA'].max()

print(f"{most_offensive_team} have scored {most_goals_scored} goals, more than any other team!")
print(f"{most_defensive_team} have conceded {most_goals_conceded} goals, more than any other team!")
final_standing
````
